```python
# 가장 주문이 많이 된 메뉴의 총 개수를 출력하시오.
# 주문서 2개,3개 -> 총 몇개가 주문이 들어왔는지 출력하시오.
# df.groupby(['item_name'])['quantity'].sum().sort_values(ascending=False).head(10)
# 막대그래프로 출력하시오.
item_graph = df.groupby(['item_name'])['quantity'].sum().head(3)
x = item_graph.index
y = item_graph.values

plt.bar(x,y)
plt.title('메뉴별 주문량')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False  # -부호 처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15   # 전체 글자폰튼크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/연령별인구현황01.xlsx',skiprows=3,nrows=1,usecols='B,E:Y',index_col='행정기관')
```

```python
# 막대그래프 출력하시오.
# df['0~4세'] = df['0~4세'].str.replace(',','').astype(int) # 1개만 변겅
```

```python
# df = df.iloc[0].str.replace(',','').astype(int)
df2 = (df.iloc[0].astype(str).str.replace(',','').astype(int))
# df2.columns
x = df2.index
y = df2.values
plt.barh(x,y)
plt.show()
# 막대 옆으로 막대그래프 출력하시오.
```

```python
# 1. x축, y축 데이터 지정
y = df.index  # 연령대 (y축)
x = df.values  # 인구수 (x축)

# 2. 그래프 그리기 (크기 설정 포함)
plt.figure(figsize=(10, 8))
plt.barh(y, x)

# 3. 제목 및 축 라벨 설정
plt.title('전국 연령대별 인구수')
plt.xlabel('인구수 (명)')
plt.ylabel('연령대')

# 4. x축 지수 표기법(1e6) 해제 및 출력
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

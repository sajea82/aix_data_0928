```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False  # -부호 처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15   # 전체 글자폰튼크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx')
# df.groupby('학년').value_counts().loc['디지털고']
school = df['학교'].value_counts(normalize=True)
explode = [0.1,0,0] # 원파트 띄우기
colors = ['#ffadad','#ffd6a5','#fdffb6','#caffbf','#9bf6ff','#a0c4ff']
school.index
school.values
plt.pie(school.values,labels=school.index,startangle=90,counterclock=False,explode=explode,\
        autopct='%.1f%%',colors=colors)
plt.show()
```
